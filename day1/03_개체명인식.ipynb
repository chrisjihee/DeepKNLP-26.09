## 4. 미션 — 글자 단위 태그를 서브워드 토큰에 맞추기

개체명인식은 앞의 둘과 **답의 단위가 다릅니다.** 문장마다 하나가 아니라 **토큰마다** 하나입니다.
그런데 문제가 하나 더 있습니다 — **정답이 붙어 있는 단위와 모델이 보는 단위가 다릅니다.**

- 원본 데이터의 정답은 **글자마다** 붙어 있습니다: `김`=`B-PS`, ` `=`O`, `씨`=`O`, …
- 모델이 보는 것은 **서브워드 토큰**입니다: `김`, `씨는`, `10`, `##개월`, …

그 둘을 잇는 코드는 **이미 다 쓰여 있습니다.** 아래 미션 셀을 읽어 보면 규칙이 그대로 보입니다.

| | 어떤 토큰인가 | 라벨 |
|---|---|---|
| 1 | `[CLS]`·`[SEP]` 같은 특수 토큰 (offset이 `(0, 0)`) | **손실에서 빼는 값** ← `m-ner-1` |
| 2 | 시작 글자의 태그가 `O` | `O` |
| 3 | 시작 글자의 태그가 `B-` 이거나, 앞 토큰과 유형이 다름 | `B-유형` (새 개체의 시작) |
| 3' | 그 밖 — 앞 토큰과 같은 유형이 이어짐 | `I-유형` |

개체명인식의 head는 토큰마다 태그 하나. 태그는 BIO 방식으로 13개입니다 — `O` 하나 + 6가지 유형 × (`B-`시작, `I-`이어짐).

여러분이 정할 것은 **세 곳뿐**입니다. 규칙을 새로 짜는 것이 아니라, 규칙에서 **핵심 조각**을 고르는 일입니다.
앞의 둘은 **학습 라벨을 만드는 쪽**, 마지막 하나는 **모델이 낸 태그를 개체로 되돌리는 쪽**입니다.

| 문항 | 정할 것 | 어디 |
|---|---|---|
| `m-ner-1` | 특수 토큰 자리에 넣을 **라벨 값** | 미션 셀 ① |
| `m-ner-2` | 토큰의 태그를 **어디서 가져올지** | 미션 셀 ① |
| `m-ner-3` | 언제 **새 개체를 시작**할지 | 미션 셀 ② |

### 빈칸 채우기

위에서 고른 답을 아래 `____` 두 곳에 옮겨 적고 셀을 실행하세요. **나머지는 그대로 둡니다.**

````python
# ══════════════════════════════════════════════════════════════════════
#  미션 m-ner-1 · m-ner-2 — 글자 단위 태그를 서브워드 토큰에 맞추기
# ══════════════════════════════════════════════════════════════════════
# 아래 코드에서 ____ **두 곳만** 채우세요. 나머지는 그대로 두면 됩니다.
# 바로 위 두 카드에서 고른 답을 그대로 옮겨 적으면 됩니다.
#
# 완성본은 `task5-llm-ft/bert_baseline.py` 에도 있습니다 — 감추지 않고 알려 드립니다.
# 먼저 스스로 골라 보고, 막히면 위 카드의 힌트를 하나씩 열어 보세요. 그래야 남는 것이 있습니다.

def build_model():
    # 토큰마다 태그 하나를 고르는 head. 태그는 BIO 13개(O + 6유형 × B/I).
    return AutoModelForTokenClassification.from_pretrained(
        MODEL, num_labels=len(L.BIO_LABELS),
        id2label=dict(enumerate(L.BIO_LABELS)), label2id=L.BIO2ID)


def encode_ner(tokenizer, rows, max_len=MAX_LEN):
    texts = [r["input"]["text"] for r in rows]
    enc = tokenizer(texts, truncation=True, max_length=max_len, return_offsets_mapping=True)
    all_labels = []
    for i, r in enumerate(rows):
        cl = r["raw"]["char_labels"]        # 글자 하나마다 "O" / "B-PS" / "I-PS" …
        labels = []
        prev_type = None
        for (s, e) in enc["offset_mapping"][i]:
            if e <= s:                     # [CLS]·[SEP] 같은 특수 토큰 — 원문 글자가 없다
                labels.append(____)   # m-ner-1
                prev_type = None
                continue
            tag = ____ if s < len(cl) else "O"      # m-ner-2 · 토큰이 시작하는 글자의 태그를 따른다
            if tag == "O":
                labels.append(L.BIO2ID["O"])
                prev_type = None
            else:
                typ = tag.split("-")[1]
                bio = "B" if (tag.startswith("B") or typ != prev_type) else "I"
                labels.append(L.BIO2ID[f"{bio}-{typ}"])
                prev_type = typ
        all_labels.append(labels)
    enc["labels"] = all_labels
    enc.pop("offset_mapping")
    return Dataset.from_dict(dict(enc))
````

### 미션 셀 ② — `m-ner-3`

여기까지가 **모델에 넣는 쪽**이었습니다. 이제 반대쪽입니다 — **모델이 낸 것을 받는 쪽**.

이 과정의 목표는 BERT · T5 · LLM 셋을 **같은 자로 재는 것**입니다.
그런데 셋의 출력이 생김새부터 다릅니다.

| | 무엇을 내놓나 |
|---|---|
| BERT | 라벨마다(또는 토큰마다) **점수**. 숫자입니다. |
| T5 · LLM | 답을 **글자로 써냅니다**. 처음부터 문자열입니다. |

그래서 BERT 쪽 숫자를 **LLM 이 내놓는 것과 같은 형식의 문자열**로 되돌립니다.
그러고 나면 셋을 **하나의 채점 함수**에 그대로 넣을 수 있습니다 — 채점을 세 벌 만들 필요가 없습니다.

위에서 고른 답을 아래 `____` **한 곳**에 옮겨 적고 셀을 실행하세요.

````python
# ══════════════════════════════════════════════════════════════════════
#  미션 m-ner-3 — 언제 새 개체를 시작할까
# ══════════════════════════════════════════════════════════════════════
# ____ **한 곳만** 채우세요. 위 카드에서 고른 답을 그대로 옮겨 적으면 됩니다.
#
# 이 함수가 하는 일: 모델이 낸 **숫자**를 LLM·T5 와 같은 형식의 **문자열**로 되돌린다.
# 셋을 같은 채점 함수로 재려면 출력 형식이 같아야 하기 때문입니다.

def merge_bio(text, offsets, pred_ids):
    """토큰마다 나온 BIO 태그를 이어 붙여 **개체 목록**으로 만든다.

    태그는 `O` / `B-PS` / `I-PS` … 처럼 생겼다. 여기서 정할 것은 하나다 —
    **언제 새 개체를 시작하고, 언제 앞 개체에 이어 붙이는가.**
    """
    ents, cur = [], None
    for (s, e), pid in zip(offsets, pred_ids):
        if e <= s:                       # [CLS]·[SEP] — 원문 글자가 없다
            continue
        tag = L.BIO_LABELS[pid]
        if tag == "O":                   # 개체가 아니다 → 모으던 것이 있으면 닫는다
            if cur:
                ents.append(cur)
            cur = None
            continue
        bio, typ = tag.split("-")        # 예: "B-PS" → bio="B", typ="PS"
        if bio == ____ or cur is None or cur["label"] != typ:   # m-ner-3
            if cur:                      # 새 개체가 시작된다 → 앞의 것을 닫고
                ents.append(cur)
            cur = {"label": typ, "start": s, "end": e}      # 새로 연다
        else:                            # 같은 유형이 이어진다 → 끝만 늘린다
            cur["end"] = e
    if cur:
        ents.append(cur)
    return [{"text": text[c["start"]:c["end"]].strip(), "label": c["label"]} for c in ents]
````

이 함수가 제대로 도는지 **작은 가짜 입력**으로 점검합니다. 모델을 돌리지 않으므로 즉시 끝납니다.

````python
# "김 씨와 이 씨" — 사람 두 명이 나란히 있는 경우. 한 덩어리로 뭉치면 안 된다.
_text = "김 씨와 이 씨"
_off = [(0, 1), (2, 3), (3, 4), (5, 6), (7, 8)]
_ids = [L.BIO2ID[t] for t in ["B-PS", "I-PS", "O", "B-PS", "I-PS"]]
_got = merge_bio(_text, _off, _ids)
assert len(_got) == 2, f"사람 두 명이어야 하는데 {len(_got)}개가 나왔습니다: {_got}"
assert all(g["label"] == "PS" for g in _got), f"유형이 PS 여야 합니다: {_got}"
print("통과 — 나란히 있는 두 사람이 따로 잡혔습니다:", _got)
````

## 8. T5 — 같은 문제를 '써내는' 방식으로

> **여기부터는 강사 시연입니다.** 학생은 셀을 그대로 실행하면서 화면을 보세요.

지금까지 BERT는 정해진 후보 중에서 **골랐습니다**. `pko-t5-base`는 다릅니다 —
지시문을 읽고 답을 **글자로 써냅니다**. 그래서 head를 바꿀 필요가 없습니다.
바뀌는 것은 모델이 아니라 **입력에 적어 주는 지시문**입니다.

여기서 쓰는 지시문과 정답 형식은 2일차에 LLM에게 줄 것과 **한 글자도 다르지 않습니다**
(`task5-llm-ft/common.py`의 `TASKS`에 한 곳에 모아 두었습니다).
같은 문장을 주어야 세 방식을 공정하게 비교할 수 있습니다.

````python
# BERT가 쓰던 GPU 메모리를 돌려준다 (나중에 다시 돌아와 이 셀만 실행해도 되게, 변수가 없으면 조용히 넘어간다)
for _n in ("model", "trainer"):
    globals().pop(_n, None)
L.free_gpu()

t5_tok, t5_model = L.load_t5()
_src, _tgt = L.to_text(train_rows[:1])
print("─ T5가 받는 입력 ──────────────────────────────")
print(_src[0])
print("─ T5가 써내야 하는 답 ─────────────────────────")
print(_tgt[0])
print("───────────────────────────────────────────────\n")

L.set_seed(42)
t5_train_ds = L.encode_t5(t5_tok, train_rows, TASK)
t5_trainer, t5_meta = L.train_t5(t5_model, t5_tok, t5_train_ds, TASK,
                                 epochs=T5_EPOCHS, lr=T5_LR, batch_size=T5_BATCH)
````

### T5 평가

BERT와 **똑같은 평가셋 300건**을 씁니다. 다만 예측을 만드는 방식이 다릅니다 —
BERT는 점수가 가장 높은 것을 한 번에 고르지만, T5는 글자를 하나씩 이어 붙여 답을 완성합니다(생성).
그래서 시간이 더 걸립니다.

````python
t5_preds = L.generate_t5(t5_model, t5_tok, eval_rows, TASK)
print("생성 예시 3건:", t5_preds[:3], "\n")

t5_summary = L.evaluate(TASK, t5_preds, eval_rows)
L.save_result(TASK, "t5", t5_summary, t5_meta)
````
